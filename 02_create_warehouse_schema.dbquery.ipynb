```sql
-- North River retail star schema: northriver_warehouse
-- One fact table + four conformed dimensions (SCD Type 2, lookup, and date)

-- Create the warehouse schema if it doesn't exist
CREATE SCHEMA IF NOT EXISTS northriver_warehouse;

-- Line-item grain fact: one row per order line item
CREATE TABLE IF NOT EXISTS northriver_warehouse.fact_orders (
  order_id      INT,
  order_item_id BIGINT,
  customer_sk   BIGINT,
  product_sk    BIGINT,
  channel_sk    BIGINT,
  date_sk       BIGINT,
  quantity      INT,
  unit_price    DECIMAL(10,2),
  revenue       DECIMAL(10,2) GENERATED ALWAYS AS (CAST(quantity * unit_price AS DECIMAL(10,2)))
);

-- Customer dimension: SCD Type 2 on region and segment
CREATE TABLE IF NOT EXISTS northriver_warehouse.dim_customer (
  customer_sk    BIGINT GENERATED ALWAYS AS IDENTITY,
  customer_id    INT,
  first_name     STRING,
  last_name      STRING,
  email          STRING,
  region         STRING,
  segment        STRING,
  effective_from DATE,
  effective_to   DATE,
  is_current     BOOLEAN
);

-- Product dimension: SCD Type 2 on category
CREATE TABLE IF NOT EXISTS northriver_warehouse.dim_product (
  product_sk     BIGINT GENERATED ALWAYS AS IDENTITY,
  product_id     INT,
  product_name   STRING,
  brand          STRING,
  category       STRING,
  subcategory    STRING,
  unit_cost      DECIMAL(10,2),
  effective_from DATE,
  effective_to   DATE,
  is_current     BOOLEAN
);

-- Channel dimension: stable lookup, no SCD
CREATE TABLE IF NOT EXISTS northriver_warehouse.dim_channel (
  channel_sk   BIGINT GENERATED ALWAYS AS IDENTITY,
  channel_name STRING NOT NULL
);

-- Date dimension: static, no SCD
CREATE TABLE IF NOT EXISTS northriver_warehouse.dim_date (
  date_sk     BIGINT,
  full_date   DATE,
  year        INT,
  quarter     INT,
  month       INT,
  day_of_week STRING
);


```
